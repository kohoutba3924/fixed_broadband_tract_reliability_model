```python
import polars as pl

# Configure Polars so Jupyter shows ALL columns and a controlled
# number of rows. This prevents truncation and keeps rendering fast.

pl.Config.set_tbl_cols(-1)   # show all columns, no matter how many
pl.Config.set_tbl_rows(10)   # show only 10 rows in the HTML output

# Load the unified tract-quarter feature dataset
df = pl.read_parquet("../../data/processed/feature_engineering/tract_quarter_features.parquet")

# Display the first 10 rows with all columns visible
df.head(10)
```

```python
print(df.shape)
```

```python
# Are any features missing or degenerate?
missing = df.null_count()
missing
```

12-14 records contain null summary stats for weighted_precipitation, 144 records have null target value (reliability_index).

```python
# check for data clearning that occured during prior data normalization, may have modeling implications
san_cols = [c for c in df.columns if "sanitization" in c]
df.filter((pl.col('lcdv2_sanitization_level') != 'none') | (pl.col('tract_sanitization_level') != 'none')).select(san_cols)
```

```python
precip_summary_cols = [
    "weighted_precipitation_mean",
    "weighted_precipitation_median",
    "weighted_precipitation_max",
    "weighted_precipitation_min",
    "weighted_precipitation_range",
    "weighted_precipitation_p10",
    "weighted_precipitation_p25",
    "weighted_precipitation_p75",
    "weighted_precipitation_p90",
    "weighted_precipitation_std",
    "weighted_precipitation_cv",
]

missing_precip = df.filter(
    pl.any_horizontal([pl.col(c).is_null() for c in precip_summary_cols])
).select(["tract", "year", "quarter", "lcdv2_sanitization_level"] + precip_summary_cols)

missing_precip
```

The null values are the result of tracts having stations that never report precipitation. Null precipitation values were converted to a value of 0 and with all 0 values the summary of such inputs results in null summary stat values. No remediation needed for modeling later.

```python
# Temporarily increase row display limit
pl.Config.set_tbl_rows(500)

missing_target_tracts = (
    df.filter(pl.col("reliability_index").is_null())
      .select(["tract"])
      .unique()
      .sort(["tract"])
)

missing_target_tracts
```

```python

# Make sure all rows and columns are displayed
pl.Config.set_tbl_cols(-1)
pl.Config.set_tbl_rows(500)

# Load the Ookla target engineering dataset
ookla = pl.read_parquet(
    "../../data/processed/feature_engineering/ookla_tract_quarter_target_features.parquet"
)

# Convert to a Python list for filtering
tract_list = missing_target_tracts["tract"].to_list()

# Filter the Ookla dataset to only those tracts
ookla_missing = (
    ookla.filter(pl.col("tract").is_in(tract_list))
         .sort(["tract", "year", "quarter"])
)

ookla_missing
```

```python
# Rows missing reliability_index in unified dataset
missing_target_full = (
    df.filter(pl.col("reliability_index").is_null())
      .select(["tract", "year", "quarter"])
      .unique()
)

# Keys present in Ookla
ookla_keys = (
    ookla.select(["tract", "year", "quarter"])
         .unique()
)

# Inner join: only rows that exist in BOTH datasets
missing_that_exist_in_ookla = (
    missing_target_full.join(
        ookla_keys,
        on=["tract", "year", "quarter"],
        how="inner"
    )
)

missing_that_exist_in_ookla
```

```python
missing_that_do_not_exist_in_ookla = (
    missing_target_full.join(
        ookla_keys,
        on=["tract", "year", "quarter"],
        how="anti"
    )
)

missing_that_do_not_exist_in_ookla
```

The 144 records with null reliability index values are the result of tract, year, quarter combinations for which no Ookla speed tests were performed, meaning no reliability_index value may be computed for those tract, year, quarter records. Updated unified_tract_quarter_candidate_features to exclude 1) tracts having a tract_code starting with 99, for those tracts that are unpopulated and 2) any records having a null reliability_index, for those tract, year, quarters without Ookla speed test data.

```python
# check for variance among numeric features, those with no variance are not of use for modeling

numeric_cols = [
    name for name, dtype in zip(df.columns, df.dtypes)
    if dtype.is_numeric()
]

variance = df.select([
    pl.var(col).alias(f"{col}_var")
    for col in numeric_cols
    if col != "reliability_index"
])
variance
```

High level variance check indicates the presence of features that are both dynamic and close to or fully static. Those indicating close to or full static ranges will likely later be dropped during feature selection, the remaining are frequent and varied enough to support robust cross-sectional modeling.
